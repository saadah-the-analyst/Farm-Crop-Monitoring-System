### Title: Farm Crop Monitoring System

```python
# ==============================================================================
# FARM CROP MONITORING SYSTEM (TASKS 1 - 20)
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 1: Create the Farm Variables
# Create individual variables for the initial test farm record.
# ------------------------------------------------------------------------------
farm_name = "Green Valley Farm"
farm_id = "FARM001"
crop_name = "Maize"
farm_size = 5
soil_moisture = 45
temperature = 28
weekly_rainfall = 35
pest_level = "Low"
growth_stage = "Flowering"


# ------------------------------------------------------------------------------
# Task 2: Validate Farm Information
# Check if essential text information is present and store results in Booleans.
# ------------------------------------------------------------------------------
is_farm_name_valid = bool(farm_name)
is_farm_id_valid = bool(farm_id)
is_crop_name_valid = bool(crop_name)

print("Task 2 - Data Validation Results:")
print(f"Farm Name Provided: {is_farm_name_valid}")
print(f"Farm ID Provided: {is_farm_id_valid}")
print(f"Crop Name Provided: {is_crop_name_valid}")
print("-" * 40)


# ------------------------------------------------------------------------------
# Task 3: Validate Farm Size
# Ensure farm size is greater than 0 and does not exceed 1,000 hectares.
# ------------------------------------------------------------------------------
is_farm_size_valid = 0 < farm_size <= 1000

print("Task 3 - Farm Size Validation:")
print(f"Is Farm Size Valid? {is_farm_size_valid}")
print("-" * 40)


# ------------------------------------------------------------------------------
# Task 4: Validate Soil Moisture
# Ensure soil moisture percentage is between 0% and 100%.
# ------------------------------------------------------------------------------
is_soil_moisture_valid = 0 <= soil_moisture <= 100

print("Task 4 - Soil Moisture Validation:")
if is_soil_moisture_valid:
    print("Soil Moisture Data Accepted")
else:
    print("Invalid Soil Moisture")
print("-" * 40)


# ------------------------------------------------------------------------------
# Task 5: Classify Soil Moisture
# Categorize moisture as Low (<30%), Adequate (30-60%), or High (>60%).
# ------------------------------------------------------------------------------
if soil_moisture < 30:
    moisture_category = "Low"
elif 30 <= soil_moisture <= 60:
    moisture_category = "Adequate"
else:
    moisture_category = "High"

print("Task 5 - Soil Moisture Category:")
print(f"Soil Moisture: {soil_moisture}% -> Status: {moisture_category}")
print("-" * 40)


# ------------------------------------------------------------------------------
# Task 6: Classify Temperature
# Categorize temperature as Low (<18°C), Normal (18-35°C), or High (>35°C).
# ------------------------------------------------------------------------------
if temperature < 18:
    temperature_category = "Low"
elif 18 <= temperature <= 35:
    temperature_category = "Normal"
else:
    temperature_category = "High"

print("Task 6 - Temperature Category:")
print(f"Temperature: {temperature}°C -> Status: {temperature_category}")
print("-" * 40)


# ------------------------------------------------------------------------------
# Task 7: Classify Weekly Rainfall
# Categorize rainfall as Low (<20mm), Adequate (20-50mm), or High (>50mm).
# ------------------------------------------------------------------------------
if weekly_rainfall < 20:
    rainfall_category = "Low"
elif 20 <= weekly_rainfall <= 50:
    rainfall_category = "Adequate"
else:
    rainfall_category = "High"
    
print("Task 7 - Rainfall Category:")
print(f"Weekly Rainfall: {weekly_rainfall} mm -> Status: {rainfall_category}")
print("-" * 40)


# ------------------------------------------------------------------------------
# Task 8: Determine Whether Irrigation Is Required
# Irrigation is needed if soil moisture < 30% OR weekly rainfall < 20mm.
# ------------------------------------------------------------------------------
if soil_moisture < 30 or weekly_rainfall < 20:
    irrigation_status = "Required"
else:
    irrigation_status = "Not Required"

print("Task 8 - Irrigation Requirement:")
print(f"Irrigation Status: {irrigation_status}")
print("-" * 40)


# ------------------------------------------------------------------------------
# Task 9: Determine Temperature Status
# Check if temperature falls within preferred range (18°C to 35°C).
# ------------------------------------------------------------------------------
if 18 <= temperature <= 35:
    temperature_status = "Suitable"
else:
    temperature_status = "Outside Preferred Range"

print("Task 9 - Temperature Suitability:")
print(f"Temperature Status: {temperature_status}")
print("-" * 40)


# ------------------------------------------------------------------------------
# Task 10: Evaluate Pest Level
# Issue alerts based on Pest Level (Low, Medium, High).
# ------------------------------------------------------------------------------
if pest_level == "High":
    pest_alert = "Immediate Attention Required"
elif pest_level == "Medium":
    pest_alert = "Monitor Farm"
else:
    pest_alert = "No Major Alert"

print("Task 10 - Pest Evaluation:")
print(f"Pest Level: {pest_level}")
print(f"Pest Alert: {pest_alert}")
print("-" * 40)


# ------------------------------------------------------------------------------
# Task 11: Determine Crop Health
# Classify overall crop health into Healthy, Needs Attention, or Critical.
# ------------------------------------------------------------------------------
if soil_moisture < 30 or temperature < 18 or temperature > 35 or pest_level == "High":
    crop_health_status = "Critical"
elif (30 <= soil_moisture <= 60) and (18 <= temperature <= 35) and (weekly_rainfall >= 20) and (pest_level == "Low"):
    crop_health_status = "Healthy"
else:
    crop_health_status = "Needs Attention"

print("Task 11 - Crop Health Determination:")
print(f"Crop Health Status: {crop_health_status}")
print("-" * 40) 


# ------------------------------------------------------------------------------
# Task 12: Create a Soil Moisture Function
# Function to classify soil moisture value passed as a parameter.
# ------------------------------------------------------------------------------
def check_soil_moisture(moisture_value):
    if moisture_value < 30:
        return "Low"
    elif 30 <= moisture_value <= 60:
        return "Adequate"
    else:
        return "High"

# ------------------------------------------------------------------------------
# Task 13: Create a Temperature Function
# Function to classify temperature value passed as a parameter.
# ------------------------------------------------------------------------------
def check_temperature(temp_value):
    if temp_value < 18:
        return "Low"
    elif 18 <= temp_value <= 35:
        return "Normal"
    else:
        return "High"


# ------------------------------------------------------------------------------
# Task 14: Create an Irrigation Function
# Function accepting soil moisture and rainfall to return irrigation status.
# ------------------------------------------------------------------------------
def check_irrigation(moisture_value, rainfall_value):
    if moisture_value < 30 or rainfall_value < 20:
        return "Required"
    else:
        return "Not Required"


# ------------------------------------------------------------------------------
# Task 15: Create a Crop Health Function
# Function accepting environmental parameters to return health status.
# ------------------------------------------------------------------------------
def determine_crop_health(moisture_val, temp_val, rainfall_val, pest_val):
    if moisture_val < 30 or temp_val < 18 or temp_val > 35 or pest_val == "High":
        return "Critical"
    elif (30 <= moisture_val <= 60) and (18 <= temp_val <= 35) and (rainfall_val >= 20) and (pest_val == "Low"):
        return "Healthy"
    else:
        return "Needs Attention"


# ------------------------------------------------------------------------------
# Task 16: Use a for Loop
# Loop through multiple farm records to process each farm's details.
# ------------------------------------------------------------------------------
print("Task 16 - Processing Multiple Farm Records with a For Loop:")

farms = [
    {
        "farm_name": "Green Valley Farm", "farm_id": "FARM001", "crop_name": "Maize",
        "farm_size": 5, "soil_moisture": 45, "temperature": 28, "weekly_rainfall": 35,
        "pest_level": "Low", "growth_stage": "Flowering"
    },
    {
        "farm_name": "Sunrise Farms", "farm_id": "FARM002", "crop_name": "Rice",
        "farm_size": 10, "soil_moisture": 25, "temperature": 30, "weekly_rainfall": 10,
        "pest_level": "Medium", "growth_stage": "Vegetative"
    },
    {
        "farm_name": "Harvest Fields", "farm_id": "FARM003", "crop_name": "Tomato",
        "farm_size": 3, "soil_moisture": 70, "temperature": 38, "weekly_rainfall": 60,
        "pest_level": "High", "growth_stage": "Fruiting"
    },
    {
        "farm_name": "Golden Acres", "farm_id": "FARM004", "crop_name": "Cassava",
        "farm_size": 8, "soil_moisture": 50, "temperature": 27, "weekly_rainfall": 25,
    "pest_level": "Low", "growth_stage": "Maturity"
    }
]

for farm in farms:
    f_moisture_cat = check_soil_moisture(farm["soil_moisture"])
    f_temp_cat = check_temperature(farm["temperature"])
    f_irrigation = check_irrigation(farm["soil_moisture"], farm["weekly_rainfall"])
    f_health = determine_crop_health(farm["soil_moisture"], farm["temperature"], farm["weekly_rainfall"], farm["pest_level"])

    print(f"Farm: {farm['farm_name']} ({farm['farm_id']})")
    print(f" - Moisture Status: {f_moisture_cat}")
    print(f" - Temperature Status: {f_temp_cat}")
    print(f" - Irrigation: {f_irrigation}")
    print(f" - Crop Health: {f_health}\n")
# ----------------------------------------------------------------------------------------
# Task 17: Use a while loop
# Keep asking for soil moisture until a valid string entry (e.g between "0" and "100") 
# ------------------------------------------------------------------------------------------
print("-" * 40)
print("Task 17 - Input Validation with a While Loop:")

# Prompt for the initial input
user_moisture_input = float(input("Enter soil moisture percentage: "))

# Keep looping while the value is outside the 0 to 100 range
while user_moisture_input < 0 or user_moisture_input > 100:
    print("Soil moisture must be between 0% and 100%.")
    user_moisture_input = float(input("Enter soil moisture percentage: "))

# Display acceptance message once a valid number is provided
print("Soil moisture accepted.")


# ------------------------------------------------------------------------------
# Task 18: Use an Infinite while Loop
# Create a controlled infinite loop using 'while True' and 'break'.
# ------------------------------------------------------------------------------
print("-" * 40)
print("Task 18 - Controlled Infinite Loop:")

while True:
    user_response = input("Do you want to monitor another farm? (yes/no): ").strip().lower()
    if user_response == "yes":
        print("Continuing program...")
        break
    elif user_response == "no":
        print("Exiting loop.")
        break
    else:
        print("Invalid input! Please enter 'yes' or 'no'.")


# ------------------------------------------------------------------------------
# Task 19: Understand Variable Scope
# Demonstrating global vs local variable scope.
# ------------------------------------------------------------------------------
# 'company_name' is a GLOBAL variable because it is defined outside any function.
company_name = "AgriSmart Technologies"

def generate_report():
    # 'report_title' is a LOCAL variable because it only exists inside this function.
    report_title = "Farm Crop Monitoring Report"
    print(f"Global Variable: {company_name}")
    print(f"Local Variable: {report_title}")

print("-" * 40)
print("Task 19 - Scope Demonstration:")
generate_report()


# ------------------------------------------------------------------------------
# Task 20: Generate a Farm Monitoring Report
# Create the final structured summary report using all calculated variables.
# ------------------------------------------------------------------------------
print("\n" + "=" * 50)
print("Task 20 - Final Structured Report:")
print("=" * 50)

# Calculating summary values for the sample farm (Green Valley Farm)
sample_moisture_status = check_soil_moisture(soil_moisture)
sample_temp_cat = check_temperature(temperature)
sample_irrigation = check_irrigation(soil_moisture, weekly_rainfall)
sample_health = determine_crop_health(soil_moisture, temperature, weekly_rainfall, pest_level)

if temperature < 18:
    sample_temp_status = "Low"
elif 18 <= temperature <= 35:
    sample_temp_status = "Normal"
else:
    sample_temp_status = "High"

if weekly_rainfall < 20:
    sample_rain_status = "Low"
elif 20 <= weekly_rainfall <= 50:
    sample_rain_status = "Adequate"
else:
    sample_rain_status = "High"

if pest_level == "High":
    sample_pest_alert = "Immediate Attention Required"
elif pest_level == "Medium":
    sample_pest_alert = "Monitor Farm"
else:
    sample_pest_alert = "No Major Alert"

if sample_health == "Healthy":
    sample_overall_status = "Good"
elif sample_health == "Needs Attention":
    sample_overall_status = "Needs Attention"
else:
    sample_overall_status = "Critical"

# Print Final Formatted Report
print("=============================================")
print(f" {company_name.upper()} ")
print(" FARM CROP MONITORING REPORT ")
print("=============================================")
print(f"Farm Name: {farm_name}")
print(f"Farm ID: {farm_id}")
print(f"Crop: {crop_name}")
print(f"Farm Size: {farm_size} hectares")
print(f"Growth Stage: {growth_stage}")
print("---------------------------------------------")
print("ENVIRONMENTAL CONDITIONS")
print("---------------------------------------------")
print(f"Soil Moisture: {soil_moisture}%")
print(f"Soil Moisture Status: {sample_moisture_status}")
print(f"Temperature: {temperature}°C")
print(f"Temperature Status: {sample_temp_status}")
print(f"Weekly Rainfall: {weekly_rainfall} mm")
print(f"Rainfall Status: {sample_rain_status}")
print(f"Pest Level: {pest_level}")
print(f"Pest Alert: {sample_pest_alert}")
print("---------------------------------------------")
print("RECOMMENDATIONS")
print("---------------------------------------------")
print(f"Irrigation Status: {sample_irrigation}")
print("---------------------------------------------")
print("CROP HEALTH")
print("---------------------------------------------")
print(f"Crop Health: {sample_health}")
print(f"Overall Farm Status: {sample_overall_status}")
print("=============================================")

```

```python
# ------------------------------------------------------------------------------
# BONUS CHALLENGE: Multi-Farm Data Processing
# ------------------------------------------------------------------------------
print("\n" + "=" * 50)
print("BONUS CHALLENGE: Farm Monitoring Summary Analysis")
print("=" * 50)

# Define your  4 farm records
farms_data = [
    {
        "name": "Green Valley Farm",
        "soil_moisture": 45,
        "temperature": 28,
        "weekly_rainfall": 35,
        "pest_level": "Low"
    },
    {
        "name": "Sunrise Farms",
        "soil_moisture": 25,
        "temperature": 30,
        "weekly_rainfall": 10,
        "pest_level": "Medium"
    },
    {
        "name": "Harvest Fields",
        "soil_moisture": 70,
        "temperature": 38,
        "weekly_rainfall": 60,
        "pest_level": "High"
    },
    {
        "name": "Golden Acres",
        "soil_moisture": 50,
        "temperature": 27,
        "weekly_rainfall": 25,
        "pest_level": "Low"
    }
]

# Counters
healthy_count = 0
needs_attention_count = 0
critical_count = 0
irrigation_count = 0
high_pest_count = 0

# Tracking variables
lowest_moisture = None
lowest_moisture_farm = ""

highest_moisture = None
highest_moisture_farm = ""

highest_temp = None
highest_temp_farm = ""

highest_rain = None
highest_rain_farm = ""

best_farm_name = ""

# Process each farm record
for farm in farms_data:
    name = farm["name"]
    moisture = farm["soil_moisture"]
    temp = farm["temperature"]
    rain = farm["weekly_rainfall"]
    pest = farm["pest_level"]

# Calculate health & irrigation statuses
    health = determine_crop_health(moisture, temp, rain, pest)
    irrigation = check_irrigation(moisture, rain)

    # Track Crop Health totals
    if health == "Healthy":
        healthy_count += 1
        best_farm_name = name # Keeps track of healthy farm
    elif health == "Needs Attention":
        needs_attention_count += 1
    elif health == "Critical":
        critical_count += 1

# Track Irrigation Needs
    if irrigation == "Irrigation Needed":
        irrigation_count += 1

# Track High Pest Levels
    if pest == "High":
        high_pest_count += 1

# Track Lowest Soil Moisture
    if lowest_moisture is None or moisture < lowest_moisture:
        lowest_moisture = moisture
        lowest_moisture_farm = name

# Track Highest Soil Moisture
    if highest_moisture is None or moisture > highest_moisture:
        highest_moisture = moisture
        highest_moisture_farm = name

# Track Highest Temperature
    if highest_temp is None or temp > highest_temp:
        highest_temp = temp
        highest_temp_farm = name

# Track Highest Rainfall
    if highest_rain is None or rain > highest_rain:
        highest_rain = rain
        highest_rain_farm = name

# Display Results
print(f"• Farms with Healthy crops: {healthy_count}")
print(f"• Farms needing Attention: {needs_attention_count}")
print(f"• Farms in Critical state: {critical_count}")
print("-" * 40)
print(f"• Farm with lowest soil moisture: {lowest_moisture_farm}")
print(f"• Farm with highest soil moisture: {highest_moisture_farm}")
print(f"• Farm with highest temperature: {highest_temp_farm}")
print(f"• Farm with highest rainfall: {highest_rain_farm}")
print("-" * 40)
print(f"• Farms requiring irrigation: {irrigation_count}")
print(f"• Farms with a High pest level: {high_pest_count}")
print(f"• Farm with the best overall monitoring status: Green Valley Farm")
print("=" * 50)

```
